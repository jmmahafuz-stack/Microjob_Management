```python
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
%matplotlib inline
```

```python
# Load the Sample Data
from pathlib import Path

csv_path = Path('sample_data.csv')
print(f"Looking for dataset at: {csv_path.resolve()}")

df = pd.read_csv(csv_path)
df.head()
```

```python
# Inspect the Data Structure
df.head()
print(f"Rows: {len(df)}")
print(f"Columns: {list(df.columns)}")
print(df.dtypes)
df.info()
```

```python
# Clean and Preprocess the Data
df = df.drop_duplicates().copy()
df = df.dropna(subset=['value'])
df['value'] = df['value'].astype(float)
print(df.isna().sum())
df.head()
```

```python
# Analyze Key Metrics
summary = df.describe().transpose()
print(summary)

if 'category' in df.columns:
    grouped = df.groupby('category')['value'].agg(['count', 'mean', 'sum'])
    print(grouped)
```

```python
# Create Visualizations
if 'category' in df.columns:
    plt.figure(figsize=(8, 5))
    df.groupby('category')['value'].sum().plot(kind='bar', color='steelblue')
    plt.title('Total Value by Category')
    plt.ylabel('Value')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    plt.figure(figsize=(8, 5))
    plt.hist(df['value'], bins=10, color='cornflowerblue', edgecolor='black')
    plt.title('Value Distribution')
    plt.xlabel('Value')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()
```

```python
# Save Processed Outputs
cleaned_path = 'cleaned_data.csv'
summary_path = 'summary_metrics.csv'

df.to_csv(cleaned_path, index=False)
summary.to_csv(summary_path)

print(f"Cleaned data saved to: {cleaned_path}")
print(f"Summary metrics saved to: {summary_path}")
```
